### load data

```python
def load_dataset_from_txt(path, word2idx, max_len=50):
    sequences = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            tokens = line.strip().split()
            if len(tokens) < 2:
                continue
            seq = encode_sentence(tokens, word2idx, max_len)
            sequences.append(seq)
    return np.array(sequences)
  
X_train_raw = load_dataset_from_txt("G:\\OneDrive - Alamein International University\\Uni stuff\\semester 6 - Spring 24-25\\NLP\\assignments\\assignment 3 v0.1\\nlp-assignment-3\\data\\train.txt", preprocessing.word2idx, max_len=50)
X_val_raw = load_dataset_from_txt("G:\\OneDrive - Alamein International University\\Uni stuff\\semester 6 - Spring 24-25\\NLP\\assignments\\assignment 3 v0.1\\nlp-assignment-3\\data\\validation.txt", preprocessing.word2idx, max_len=50)
X_test_raw = load_dataset_from_txt("G:\\OneDrive - Alamein International University\\Uni stuff\\semester 6 - Spring 24-25\\NLP\\assignments\\assignment 3 v0.1\\nlp-assignment-3\\data\\test.txt", preprocessing.word2idx, max_len=50)

X_train = X_train_raw[:, :]
# Pad sequences to length 50
y_train = X_train_raw[:, :]
y_train = np.expand_dims(y_train, -1)
##### Trial 1: default parameters ######
model = build_bi_lstm_model(vocab_size=len(preprocessing.word2idx))
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5, batch_size=256)
model.summary()
```
